1. Array Creation Function

```python
import numpy as np
```

```python
#creat an array from a list
a= np.array([1,2,3])
print('Array a:', a)
```

```python
#creat an array with evenly spaced values
b= np.arange(0,10,2) #values from 0 to 10 with step 2
print('Array b:', b)
```

```python
#Creat an array with linearly spaced values
c= np.linspace(0,1,5)
print('Array c:', c)
```

```python
#Creat an array filled with zeros
d= np.zeros((2,3))
print('Array d:\n', d)
```

```python
#creat an array filled with ones
e= np.ones((3,2))
print('Array e:\n', e)
```

```python
#creat an identity matrix
f= np.eye(4) #4x4 identity matrix
print('identity matrix f:\n', f)
```

2. Array Manipulation Functions

```python
#Reshape an array
a1 = np.array([1,2,3])
reshaped = np.reshape(a1, (1,3))
print('Reshaped array:', reshaped)
```

```python
#Flatten an array
f1= np.array([[1,2],[3,4]])
flattened = np.ravel(f1) 
print('Flattened array:', flattened)
```

```python
#Transpose an array
e1= np.array([[1,2],[3,4]])
transposed = np.transpose(e1)
print('Transposed array:\n', transposed)
```

```python
#Stack arrays vertically
a2 = np.array([1,2])
b2= np.array([3,4])
stacked = np.vstack([a2,b2]) #vertical staking
print('Vertical stacked arrays:\n', stacked)
```

```python
stacked1= np.hstack([a2,b2]) #horizontal stacked
print('Horizontal stacked arrays:\n', stacked1)
```

```python
#Splitting an array
a3= np.arange(0,12).reshape(3,4)
split= np.hsplit(a3,2) #horizontal splitting
print('Horizontal splitted arrays:\n', split)
```

```python
split1= np.vsplit(a3,3)
print('Vertical splitted arrays:\n', split1)
```

4. Statistical Functions

```python
s= np.array([1,2,3,4])
mean= np.mean(s)
print('Mean of s:', mean)
```

```python
#Standard deviation of an array
std_dev= np.std(s)
print('Standard deviation of s:', std_dev)
```

```python
#Minimum element of an array
minimum = np.min(s)
print('Minimum of s:', minimum)
```

```python
#Maximum element of an array
maximum= np.max(s)
print('Maximum of s:', maximum)
```

5. Linear Algebra Function

```python
#Creat a matrix
matrix = np.array([[1,2],[3,4]])
```

```python
#Determinant of a matrix
determinant = np.linalg.det(matrix)
print('Determinant of matrix:', determinant)
```

```python
#Inverse of a matrix
inverse = np.linalg.inv(matrix)
print('Inverse of matrix;\n', inverse)
```

6. Random Sampling Functions

7. Boolean and Logical Functions

```python
# Check if all the elements are True
logical_test = np.array([True, False, True])
all_true = np.all(logical_test) #check if all are true
print('All elements True:', all_true)
```

```python
# Check if all elements are True
logical_test = np.array([False, False, False])
all_true = np.all(logical_test)  # Check if all are True
print("All elements True:", all_true)
```

```python
# Check if any elements are True
# any
logical_test = np.array([True, False, True])
any_true = np.any(logical_test)  # Check if any are True
print("Any elements True:", any_true)
```

8. Set Operations

```python
# Intersection of two arrays
set_a = np.array([1,2,3,4])
set_b = np.array([3,4,5,6])
intersection = np.intersect1d(set_a, set_b)
print('Intersection of a and b:', intersection)
```

```python
# Union of two arrays
union= np.union1d(set_a, set_b)
print('Union of a and b:',union )
```

```python
set_a = np.arange(6).reshape(2,3)
set_b = np.arange(6,12).reshape(2,3)
union= np.union1d(set_a, set_b)
print('Union of a and b:', union)
```

```python
set_a = np.arange(6).reshape(2,3)
set_b = np.arange(4,10).reshape(2,3)
intersection = np.intersect1d(set_a, set_b)
print('intersection of a and b:', intersection)
```

9. Array Attribute Functions

```python
# Array attributes
a = np.array([1, 2, 3])
shape = a.shape  # Shape of the array
size = a.size    # Number of elements
dimensions = a.ndim  # Number of dimensions
dtype = a.dtype   # Data type of the array

print("Shape of a:", shape)
print("Size of a:", size)
print("Number of dimensions of a:", dimensions)
print("Data type of a:", dtype)
```

10. Other Functions

```python
# Create a copy of an array
a = np.array([1, 2, 3])
copied_array = np.copy(a) 
print("Copied array:", copied_array)
```

```python
# Size in bytes of an array
array_size_in_bytes = a.nbytes
print("Size of a in bytes:", array_size_in_bytes)
```

```python
# Check if two arrays share memory
shared = np.shares_memory(a, copied_array)
print("Do a and copied_array share memory?", shared)
```
